```python
from google.colab import drive
drive.mount('/content/drive')

import os, zipfile, requests
from pathlib import Path

RAW = '/content/drive/MyDrive/TrafficAI_ControlCenter/datasets/raw'
gtfs_dir = Path(f'{RAW}/delhi_gtfs')
gtfs_dir.mkdir(parents=True, exist_ok=True)

# Working archive mirror (Delhi bus GTFS)
url = 'https://raw.githubusercontent.com/transitland/gtfs-archives-not-hosted-elsewhere/3db26c0092b6efeb1886a99b4fc0765122b0282b/delhi-bus.zip'
zip_path = gtfs_dir / 'delhi_bus_gtfs.zip'

print('Downloading Delhi GTFS...')
r = requests.get(url, timeout=300)
r.raise_for_status()
zip_path.write_bytes(r.content)
print('Saved:', zip_path, 'bytes:', zip_path.stat().st_size)

# Extract
out = gtfs_dir / 'extracted'
out.mkdir(exist_ok=True)
with zipfile.ZipFile(zip_path) as z:
    z.extractall(out)

needed = ['stops.txt', 'routes.txt', 'trips.txt', 'stop_times.txt']
found = {n: any(p.name == n for p in out.rglob(n)) for n in needed}
print('Files found:', found)
print('2.2 OK' if all(found.values()) else '2.2 INCOMPLETE')

# Show a few files
for p in sorted(out.rglob('*.txt'))[:20]:
    print(' -', p.relative_to(out))
```

```python
from google.colab import drive
drive.mount('/content/drive')

from pathlib import Path
import requests

RAW = Path('/content/drive/MyDrive/TrafficAI_ControlCenter/datasets/raw')
ngsim = RAW / 'ngsim'
ngsim.mkdir(parents=True, exist_ok=True)

url = 'https://data.transportation.gov/api/v3/views/8ect-6jqj/export.csv?accessType=DOWNLOAD'
out = ngsim / 'ngsim_all.csv'

print('Downloading NGSIM CSV (can be large / slow)...')
try:
    with requests.get(url, stream=True, timeout=600) as r:
        r.raise_for_status()
        with open(out, 'wb') as f:
            for chunk in r.iter_content(chunk_size=1024*1024):
                if chunk:
                    f.write(chunk)
    print('Saved:', out, 'size_MB:', round(out.stat().st_size/1e6, 1))
except Exception as e:
    print('Auto-download failed:', e)
    print('Manual backup: open this page and export CSV/zip into datasets/raw/ngsim/')
    print('https://catalog.data.gov/dataset/next-generation-simulation-ngsim-vehicle-trajectories-and-supporting-data')
```

```python
from pathlib import Path
RAW = Path('/content/drive/MyDrive/TrafficAI_ControlCenter/datasets/raw')

def ok(folder, exts=None):
    files = [p for p in (RAW/folder).rglob('*') if p.is_file()]
    if exts:
        files = [p for p in files if p.suffix.lower() in exts]
    return len(files)

checks = {
    'india_detection': ok('india_detection') > 0,
    'delhi_gtfs': ok('delhi_gtfs', ['.txt', '.zip']) > 0 or ok('delhi_gtfs') > 0,
    'delhi_bus_gps': ok('delhi_bus_gps', ['.xlsx', '.zip', '.csv']) > 0,
    'aicity2021_t4': ok('aicity2021_t4', ['.mp4']) > 0,
    'ngsim': ok('ngsim') > 0,
    'osm_india': ok('osm_india') > 0,
    'uturn_request_note': (RAW/'manual_requests'/'UTURN_TODO.txt').exists(),
}
for k,v in checks.items():
    print(('OK ' if v else 'MISSING '), k)

must = ['india_detection','delhi_gtfs','delhi_bus_gps','aicity2021_t4']
should = ['ngsim','osm_india','uturn_request_note']
print('\nMUST ready:', all(checks[m] for m in must))
print('SHOULD ready:', all(checks[s] for s in should))
print('STEP 2 FULL OK' if all(checks.values()) else 'STEP 2 mostly ready — fix MISSING')
```
